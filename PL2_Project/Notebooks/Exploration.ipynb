```python
import pandas as pd
```

```python
races=pd.read_csv('../Data/races.csv')
drivers=pd.read_csv('../Data/drivers.csv')
results=pd.read_csv('../Data/results.csv')
qualifying=pd.read_csv('../Data/qualifying.csv')
pit_stops=pd.read_csv('../Data/pit_stops.csv')
lap_time=pd.read_csv('../Data/lap_times.csv')
constructors=pd.read_csv('../Data/constructors.csv')
```

```python
dataset={
    "races":races,
    "drivers":drivers,
    "results":results,
    "qualifying":qualifying,
    "pit_stops":pit_stops,
    "lap_time":lap_time,
    "constructors":constructors
}
```

```python
master=results.merge(
    races[['raceId','year','name']],
    on='raceId'
)
master=master.merge(
    drivers[['driverId','forename','surname']],
    on='driverId'
)
master=master.merge(
    constructors[['constructorId','name']],
    on='constructorId'
)
```

```python
master.rename(
    columns={
        'name_x':'race_name',
        'name_y':'constructor_name'
    },
    inplace=True
)
```

```python
master['driver_name']=(
    master['forename']
    +" "
    +master['surname']
)
```

```python
master.to_csv(
    '../Data/master.csv',
    index=False,
)
```
